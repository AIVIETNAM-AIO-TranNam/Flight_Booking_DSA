```python
import pandas as pd
```

```python
users = pd.read_csv("users.csv")
users.loc[users["role"] == "staff", "role"] = "promotional"
users.loc[users["role"] == "admin", "role"] = "economy"
users.to_csv("users.csv", index=False)
users["role"].value_counts()
```

```python
users = pd.read_csv("users.csv")
priority_map = {
    "promotional": 10,
    "economy": 20,
    "flexible": 30,
    "business": 40
}
users["priority_score"] = users["role"].map(priority_map).fillna(0).astype(int)
users.to_csv("users.csv", index=False)
users["priority_score"].value_counts()
```
